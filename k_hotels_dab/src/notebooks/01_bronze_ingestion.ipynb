## INCREMENTAL BRONZE INGESTION

```python
dbutils.widgets.text("source","")
```

```python
source_value = dbutils.widgets.get("source")
source_value
```

```python
df = spark.readStream.format("cloudFiles")\
        .option("cloudFiles.format", "csv")\
        .option("cloudFiles.schemaLocation", f"/Volumes/hotel_project/bronze/bronzevolume/{source_value}/checkpoints/")\
        .option("cloudFiles.schemaEvolutionMode", "rescue")\
        .load(f"/Volumes/hotel_project/raw_data/raw_files/{source_value}/")
```

```python
df.writeStream.format("delta")\
    .outputMode("append")\
    .trigger(once=True)\
    .option("checkpointLocation", f"/Volumes/hotel_project/bronze/bronzevolume/{source_value}/checkpoints/")\
    .option("path", f"/Volumes/hotel_project/bronze/bronzevolume/{source_value}/data/")\
    .start()
```
